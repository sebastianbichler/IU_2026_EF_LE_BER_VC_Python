# Beispiel für die Verwendung statischer Typisierung

### Ausführung der Check für statische Typisierung in der Datei mit definierten Typen

```python
! mypy ./src/models/player_model.py ./src/services/player_service.py
```

Da die Dateien Type-Definitionen enthalten, ist klar, dass es in den Service-Methoden keine potenziellen Type-Fehler geben kann.

### Ausführung der Check für statische Typisierung in der Datei ohne definierte Typen

```python
! mypy ./src/models/game_model.py ./src/services/game_service.py
```

Da die Dateien keine Type-Definitionen enthalten, könnten potenzielle Type-Bugs oder Fehler nicht nur in den Service-Methoden, sondern auch im Modell auftreten, beispielsweise aufgrund falscher Typen zwischen dem Modell in der Anwendung und der Entität in der Datenbank.
